3. Calculo da área de um circulo

```python
vr = float(input("Digite o valor do raío: "));
pi = 3.14;
area = pi*vr*vr;
print("A área do circulo é: ",area);
```

4. Conversão de temperatura C° para F°

```python
t = float(input("Digite a temperatura em Celsius: "));
f = (9*t/5) + 32;
print(t,"C° em F° é igual a: F°",f)
```

9. Fatorial de um número fornecido

```python
n = int(input("Digite um número: "));
f = 1;
for i in range(1,n+1):
  f = f*i;
print(n,"Fatorial é:",f);
```

10. Exibe os 10 primeiros números da sequência de fibonacci

```python
n = 10;
a,b = 0,1;
for i in range(n):
  print(a,end=" ")
  a,b = b,a+b;
```
